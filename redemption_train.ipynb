```python

print("STEP 1: Loading environment", flush=True)

from dotenv import load_dotenv
import os
from huggingface_hub import login
from pricer.evaluator import evaluate
from pricer.deep_neural_network import DeepNeuralNetworkRunner
from pricer.items import Item

load_dotenv(override=True)
hf_token = os.environ['HF_API_KEY']
login(hf_token, add_to_git_credential=True)

print("STEP 2: Loading dataset", flush=True)

username = "ed-donner"
dataset = f"{username}/items_full"

train, val, test = Item.from_hub(dataset)

print(f"Loaded {len(train):,} training items, "
      f"{len(val):,} validation items, "
      f"{len(test):,} test items", flush=True)

print("STEP 3: Creating runner", flush=True)

runner = DeepNeuralNetworkRunner(train, val[:1000])

print("STEP 4: Running setup", flush=True)

runner.setup()
# runner.setup_for_inference()

print("STEP 5: Loading checkpoint", flush=True)

runner.load('deep_neural_network_epoch_2_batchsize_6400.pth', 'mps')

# print("Testing direct inference", flush=True)

# result = runner.inference(test[0])

# print(f"Success: {result}", flush=True)

print("STEP 6: Starting evaluation", flush=True)

def deep_neural_network(item):
    return runner.inference(item)

evaluate(deep_neural_network, test, workers=1)

print("STEP 7: Completed", flush=True)
```
